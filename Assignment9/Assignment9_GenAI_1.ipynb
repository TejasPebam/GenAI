# Assignment 9: Variational Autoencoder for Compression and Denoising

**Objective:** This assignment applies a Variational Autoencoder (VAE) to two tasks: reducing the representation size of MNIST images and removing added noise. It also examines how the choice of latent-space size influences reconstruction and denoising performance.

## Step 1: Importing the Required Libraries

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import numpy as np
```

## Step 2: Setting Parameters and Loading the MNIST Dataset

The experiment follows the main settings used previously: a batch size of 128, 20 training epochs, and a learning rate of 0.001. An additional noise setting is used later for the denoising experiment. MNIST images are loaded with pixel values normalized to the [0, 1] range.

```python
# Basic parameters (same as Assignment 8)
BATCH_SIZE = 128
EPOCHS = 20
LEARNING_RATE = 0.001

# Extra settings for this assignment
NOISE_STD = 0.3          # standard deviation of the Gaussian noise used in the denoising part
NUM_TEST_IMAGES = 10     # number of test images used in Parts B and C

# Fixing the seed so the results can be repeated
torch.manual_seed(42)
np.random.seed(42)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")
```

```python
transform = transforms.Compose([
    transforms.ToTensor()      # pixel values stay in [0, 1], same as Assignment 8
])

train_dataset = torchvision.datasets.MNIST(
    root='./data',
    train=True,
    download=True,
    transform=transform
)

test_dataset = torchvision.datasets.MNIST(
    root='./data',
    train=False,
    download=True,
    transform=transform
)

train_loader = torch.utils.data.DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True
)

test_loader = torch.utils.data.DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False
)

print(f"Training samples : {len(train_dataset)}")
print(f"Testing samples  : {len(test_dataset)}")
```

## Step 3: VAE Architecture and Loss Function

The VAE follows the same basic encoder-decoder structure used earlier (784 → 256 → 64 → latent → 64 → 256 → 784). The latent dimension is made configurable through `latent_dim`, allowing the same VAE implementation to be tested with 2, 16, and 32 latent values. The loss calculation remains unchanged.

```python
class VAE(nn.Module):
    def __init__(self, latent_dim=2):
        super(VAE, self).__init__()

        self.latent_dim = latent_dim

        # Encoder
        self.encoder = nn.Sequential(
            nn.Linear(784, 256),
            nn.ReLU(),
            nn.Linear(256, 64),
            nn.ReLU()
        )

        # Mean and log variance layers
        self.fc_mu      = nn.Linear(64, latent_dim)
        self.fc_log_var = nn.Linear(64, latent_dim)

        # Decoder
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 256),
            nn.ReLU(),
            nn.Linear(256, 784),
            nn.Sigmoid()
        )

    def reparameterize(self, mu, log_var):
        std = torch.exp(0.5 * log_var)
        eps = torch.randn_like(std)
        return mu + eps * std

    def forward(self, x):
        x = x.view(-1, 784)

        # Encode
        h       = self.encoder(x)
        mu      = self.fc_mu(h)
        log_var = self.fc_log_var(h)

        # Sample from the distribution
        z = self.reparameterize(mu, log_var)

        # Decode
        output = self.decoder(z)

        return output, mu, log_var
```

```python
def vae_loss(output, target, mu, log_var):

    # Reconstruction loss (MSE)
    target = target.view(-1, 784)
    reconstruction_loss = nn.functional.mse_loss(output, target, reduction='sum')

    # KL Divergence loss
    kl_loss = -0.5 * torch.sum(1 + log_var - mu.pow(2) - log_var.exp())

    # Total loss
    total_loss = reconstruction_loss + kl_loss

    return total_loss, reconstruction_loss, kl_loss


# Checking the 2-dimensional VAE (the one from Assignment 8)
vae_2 = VAE(latent_dim=2).to(device)
print(vae_2)
```

## Step 4: Training the VAE with a 2-Dimensional Latent Space

The training procedure is wrapped inside a function so that it can be reused for the larger latent spaces in Part D. First, the VAE with a latent size of 2 is trained for 20 epochs using the selected learning rate and batch size.

```python
def train_vae(model, epochs=EPOCHS, lr=LEARNING_RATE):

    optimizer = optim.Adam(model.parameters(), lr=lr)

    recon_losses = []
    kl_losses    = []
    total_losses = []

    for epoch in range(1, epochs + 1):

        total_loss_epoch = 0
        recon_loss_epoch = 0
        kl_loss_epoch    = 0

        for images, _ in train_loader:
            images = images.to(device)

            output, mu, log_var = model(images)

            total_loss, recon_loss, kl_loss = vae_loss(output, images, mu, log_var)

            optimizer.zero_grad()
            total_loss.backward()
            optimizer.step()

            total_loss_epoch += total_loss.item()
            recon_loss_epoch += recon_loss.item()
            kl_loss_epoch    += kl_loss.item()

        avg_total = total_loss_epoch / len(train_loader)
        avg_recon = recon_loss_epoch / len(train_loader)
        avg_kl    = kl_loss_epoch    / len(train_loader)

        total_losses.append(avg_total)
        recon_losses.append(avg_recon)
        kl_losses.append(avg_kl)

        print(f"Epoch [{epoch}/{epochs}]  Total Loss: {avg_total:.2f}  Reconstruction Loss: {avg_recon:.2f}  KL Loss: {avg_kl:.2f}")

    return total_losses, recon_losses, kl_losses


# Train the 2-dimensional VAE (the Assignment 8 model)
total_2, recon_2, kl_2 = train_vae(vae_2)
```

## Step 5: Part A - Compression

Each MNIST image contains 784 pixel values because its dimensions are 28 × 28. The encoder maps these values into a much smaller latent vector. The compression ratio is obtained by comparing the original number of values with the number stored in the latent representation.

**Note:** The encoder produces a mean (`mu`) and a log-variance vector. For the deterministic representation used in this experiment, the mean is used as the encoded representation, so its length is treated as the latent size.

```python
ORIGINAL_SIZE = 784      # 28 x 28 pixel values

def compression_ratio(latent_dim):
    return ORIGINAL_SIZE / latent_dim

latent_dim = 2
print(f"Original image size     : {ORIGINAL_SIZE} values")
print(f"Latent representation   : {latent_dim} values")
print(f"Compression ratio       : {ORIGINAL_SIZE} / {latent_dim} = {compression_ratio(latent_dim):.0f} : 1")

# Same comparison in bits (MNIST pixels are stored as 8-bit numbers, the latent values are 32-bit floats)
original_bits = ORIGINAL_SIZE * 8
latent_bits   = latent_dim * 32
print(f"\nIn bits: {original_bits} bits vs {latent_bits} bits = {original_bits / latent_bits:.0f} : 1")
```

The 392 : 1 figure represents the reduction in the number of stored values. If numerical precision is also considered, the original pixels use 8-bit values while the latent representation uses 32-bit floating-point values, giving a smaller effective bit-level saving of 98 : 1. The process is lossy because the decoder reconstructs an approximation rather than the exact input image.

## Step 6: Part B - Image Reconstruction

Ten test images are passed through the encoder and then reconstructed by the decoder. Each reconstructed image is compared with its original version using Mean Squared Error (MSE). The encoder mean (`mu`) is used as the latent representation, which keeps the reconstruction deterministic for the same input.

```python
def encode(model, images):
    # returns the latent code (the mean mu) of the images
    model.eval()
    with torch.no_grad():
        h  = model.encoder(images.view(-1, 784))
        mu = model.fc_mu(h)
    return mu

def decode(model, z):
    # returns the images rebuilt from the latent codes
    model.eval()
    with torch.no_grad():
        return model.decoder(z)

# the first 10 test images (the same ones used in Assignment 8, Step 11)
test_images, test_labels = next(iter(test_loader))
sample_images = test_images[:NUM_TEST_IMAGES].to(device)
sample_labels = test_labels[:NUM_TEST_IMAGES]

# 1. Encode  2. Decode
latent_codes  = encode(vae_2, sample_images)
reconstructed = decode(vae_2, latent_codes)

# 4. MSE for each image (average of the squared pixel differences over the 784 pixels)
original_flat = sample_images.view(-1, 784)
mse_per_image = ((reconstructed - original_flat) ** 2).mean(dim=1)

print("Image  Digit  Latent code (2 values)      MSE")
print("-" * 50)
for i in range(NUM_TEST_IMAGES):
    z = latent_codes[i].cpu().numpy()
    print(f"{i+1:>5}  {sample_labels[i].item():>5}  [{z[0]:>7.3f}, {z[1]:>7.3f}]      {mse_per_image[i].item():.4f}")

print("-" * 50)
print(f"Average MSE over the 10 images: {mse_per_image.mean().item():.4f}")
```

The figure compares the original images in the top row with their reconstructed versions below. The MSE for each sample is displayed above the corresponding reconstruction.

```python
originals = sample_images.cpu().view(-1, 28, 28).numpy()
rebuilt   = reconstructed.cpu().view(-1, 28, 28).numpy()

plt.figure(figsize=(15, 3.6))
for i in range(NUM_TEST_IMAGES):

    # Original images - top row
    plt.subplot(2, NUM_TEST_IMAGES, i + 1)
    plt.imshow(originals[i], cmap='gray', vmin=0, vmax=1)
    plt.title(f'Digit {sample_labels[i].item()}', fontsize=9)
    plt.axis('off')

    # Reconstructed images - bottom row
    plt.subplot(2, NUM_TEST_IMAGES, i + 1 + NUM_TEST_IMAGES)
    plt.imshow(rebuilt[i], cmap='gray', vmin=0, vmax=1)
    plt.title(f'MSE {mse_per_image[i].item():.3f}', fontsize=9)
    plt.axis('off')

plt.suptitle('Original (top) vs Reconstructed (bottom), 2-Dimensional VAE')
plt.tight_layout()
plt.show()
```

Because ten images alone may not represent the overall model performance, the reconstruction MSE is additionally evaluated across the complete set of 10,000 test images for the final comparison.

```python
def test_set_mse(model):
    # average MSE per pixel over the whole test set, using the mean (mu) as the latent code
    model.eval()
    total_error = 0.0
    count = 0
    with torch.no_grad():
        for images, _ in test_loader:
            images = images.to(device)
            flat   = images.view(-1, 784)
            mu     = model.fc_mu(model.encoder(flat))
            recon  = model.decoder(mu)
            total_error += ((recon - flat) ** 2).sum().item()
            count += flat.numel()
    return total_error / count

mse_full_2 = test_set_mse(vae_2)
print(f"MSE over all 10,000 test images (2-dimensional VAE): {mse_full_2:.4f}")
```

## Step 7: Part C - Image Denoising

Gaussian noise is added to the selected 10 test images and the noisy versions are processed through the trained 2-dimensional VAE. The clean image, noisy input, and VAE output are then viewed together. A noise standard deviation of 0.3 is used, and the resulting pixel values are limited to the [0, 1] interval.

```python
def add_noise(images, std=NOISE_STD, seed=0):
    # adds Gaussian noise with a fixed seed, so the same noise can be reused for every model
    g = torch.Generator(device=images.device)
    g.manual_seed(seed)
    noise = torch.randn(images.shape, generator=g, device=images.device) * std
    return torch.clamp(images + noise, 0.0, 1.0)

# noisy version of the same 10 test images
noisy_sample = add_noise(sample_images)

# pass the noisy images through the VAE (encode, then decode)
noisy_codes = encode(vae_2, noisy_sample)
denoised    = decode(vae_2, noisy_codes)

# errors measured against the CLEAN images
clean_flat     = sample_images.view(-1, 784)
mse_noisy      = ((noisy_sample.view(-1, 784) - clean_flat) ** 2).mean(dim=1)
mse_denoised   = ((denoised - clean_flat) ** 2).mean(dim=1)

print("Image  Digit  Noisy vs clean  Denoised vs clean  Clean recon. vs clean")
print("-" * 70)
for i in range(NUM_TEST_IMAGES):
    print(f"{i+1:>5}  {sample_labels[i].item():>5}  {mse_noisy[i].item():>13.4f}  {mse_denoised[i].item():>16.4f}  {mse_per_image[i].item():>20.4f}")

print("-" * 70)
print(f"Average  {'':>5}  {mse_noisy.mean().item():>13.4f}  {mse_denoised.mean().item():>16.4f}  {mse_per_image.mean().item():>20.4f}")
```

```python
clean_imgs    = sample_images.cpu().view(-1, 28, 28).numpy()
noisy_imgs    = noisy_sample.cpu().view(-1, 28, 28).numpy()
denoised_imgs = denoised.cpu().view(-1, 28, 28).numpy()

rows = [(clean_imgs, 'Clean'), (noisy_imgs, 'Noisy'), (denoised_imgs, 'Denoised')]

plt.figure(figsize=(15, 5.2))
for r, (imgs, name) in enumerate(rows):
    for i in range(NUM_TEST_IMAGES):
        plt.subplot(3, NUM_TEST_IMAGES, r * NUM_TEST_IMAGES + i + 1)
        plt.imshow(imgs[i], cmap='gray', vmin=0, vmax=1)
        plt.axis('off')
        if i == 0:
            plt.title(name, fontsize=10, loc='left')

plt.suptitle('Clean (top), Noisy (middle) and Denoised (bottom), 2-Dimensional VAE')
plt.tight_layout()
plt.show()
```

The denoising calculation is repeated for the complete 10,000-image test set. The identical noise setting is applied across the models so that their denoising results can be compared under the same conditions.

```python
def test_set_denoise_mse(model, std=NOISE_STD):
    # average MSE per pixel over the whole test set, measured against the clean images
    model.eval()
    g = torch.Generator(device=device)
    g.manual_seed(123)

    err_noisy = 0.0
    err_denoised = 0.0
    count = 0
    with torch.no_grad():
        for images, _ in test_loader:
            flat  = images.to(device).view(-1, 784)
            noisy = torch.clamp(flat + torch.randn(flat.shape, generator=g, device=device) * std, 0.0, 1.0)
            recon = model.decoder(model.fc_mu(model.encoder(noisy)))
            err_noisy    += ((noisy - flat) ** 2).sum().item()
            err_denoised += ((recon - flat) ** 2).sum().item()
            count += flat.numel()
    return err_noisy / count, err_denoised / count

noisy_mse_full, denoised_mse_full_2 = test_set_denoise_mse(vae_2)
print(f"Noisy image vs clean image    : {noisy_mse_full:.4f}")
print(f"Denoised output vs clean image: {denoised_mse_full_2:.4f}")
```

## Step 8: Part D - Testing Larger Latent Dimensions

The latent representation is expanded from 2 values to 16 and then 32 values. All other important training settings, including the architecture, loss, epochs, and learning rate, remain unchanged. This makes the latent dimension the main variable being investigated.

```python
# 16-dimensional VAE
torch.manual_seed(42)
vae_16 = VAE(latent_dim=16).to(device)
total_16, recon_16, kl_16 = train_vae(vae_16)
```

```python
# 32-dimensional VAE
torch.manual_seed(42)
vae_32 = VAE(latent_dim=32).to(device)
total_32, recon_32, kl_32 = train_vae(vae_32)
```

The value-based compression ratios are obtained from 784 divided by the corresponding latent dimension. Since the original pixels are 8-bit values and the latent variables are 32-bit floats, the corresponding bit-level ratios are 98 : 1, 12.25 : 1, and 6.125 : 1 for latent sizes 2, 16, and 32.

## Written Explanation

**Part A: Compression.** An MNIST image contains 784 values. With a 2-value latent representation, the value-count compression ratio is 392 : 1. Increasing the latent size gives ratios of 49 : 1 for 16 dimensions and 24.5 : 1 for 32 dimensions. Since the reconstructed image is only an approximation of the input, this is lossy compression.

**Part B: Reconstruction.** The 2-dimensional model produced an average MSE of 0.0378 and its reconstructed images were relatively blurry. In the selected examples, four digits changed identity; for instance, the 2 was reconstructed as a 3 and the 4s resembled 9s. Pixel-wise MSE alone does not always indicate whether the digit identity is correct. For example, the incorrectly reconstructed 4-to-9 sample had an MSE of 0.026, while a correctly reconstructed 9 had an MSE of 0.057.

**Part C: Denoising.** Gaussian noise was introduced into the test images before they were processed by the 2-dimensional VAE. Although the generated images appeared smoother, several outputs did not preserve the original digit identity. The denoising MSE was 0.0670, which is higher than the 0.0467 error of the noisy input. This means that the model did not provide effective denoising in this experiment. One possible explanation is that the VAE was trained using clean images rather than noisy training examples.

**Part D: Changing the latent dimension.** Increasing the latent size from 2 to 16 produced a substantial reconstruction improvement, reducing the MSE from 0.0378 to 0.0202. Increasing it further to 32 produced only a small additional reduction to 0.0191. Therefore, the main improvement occurred when moving from 2 to 16 dimensions, while the extra capacity of 32 dimensions gave only a limited benefit. The trade-off is that larger latent spaces provide less compression.

For denoising, the MSE values decreased from 0.0670 to 0.0591 and then to 0.0468 as the latent dimension increased. However, the denoising quality still remained limited. The 16-dimensional results contained many 8-like shapes, while the 32-dimensional model preserved some parts of the original 7 and 0 but still produced noticeable smudging. The 32-dimensional error was approximately equal to that of the noisy input.

**Limitations.** Each latent-size model was trained only once, and the denoising experiment used a single Gaussian noise level of 0.3. The visual examples contain only 10 images, so the complete 10,000-image test-set results are more representative of the overall performance.
