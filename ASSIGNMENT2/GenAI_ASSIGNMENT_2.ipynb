# ASSIGNMENT 2: Exploring Text Generation Using GPT-2

## Objective
The aim of this assignment is to explore how different text-generation parameters affect the output produced by the pretrained GPT-2 language model.

This notebook carries out five experiments by changing generation parameters and then produces three versions using a custom prompt.

```python
# Install required libraries (run this cell once)
!pip install -q transformers torch
```

```python
import torch
from transformers import pipeline, set_seed
```

```python
# Load the pretrained GPT-2 model
generator = pipeline(
    "text-generation",
    model="openai-community/gpt2",
    pad_token_id=50256
)

# Set seed for reproducible results
set_seed(42)

print("GPT-2 model loaded successfully!")
```

```python
prompt = "Artificial intelligence is changing the way we"

print("Original Prompt:")
print(prompt)
```

```python
def generate_text(prompt, temperature=1.0, top_k=50, top_p=1.0, repetition_penalty=1.0):

    output = generator(
        prompt,
        max_new_tokens=60,
        do_sample=True,
        temperature=temperature,
        top_k=top_k,
        top_p=top_p,
        repetition_penalty=repetition_penalty,
        num_return_sequences=1,
        pad_token_id=50256
    )

    return output[0]["generated_text"]
```

```python
set_seed(42)

output_1 = generate_text(
    prompt,
    temperature=0.3,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.0
)

print("EXPERIMENT 1")
print("Parameter changed: Temperature")
print("Value: 0.3")
print()
print(output_1)
```

```python
set_seed(42)

output_2 = generate_text(
    prompt,
    temperature=1.2,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.0
)

print("EXPERIMENT 2")
print("Parameter changed: Temperature")
print("Value: 1.2")
print()
print(output_2)
```

```python
set_seed(42)

output_3 = generate_text(
    prompt,
    temperature=1.0,
    top_k=10,
    top_p=1.0,
    repetition_penalty=1.0
)

print("EXPERIMENT 3")
print("Parameter changed: Top-k")
print("Value: 10")
print()
print(output_3)
```

```python
set_seed(42)

output_4 = generate_text(
    prompt,
    temperature=1.0,
    top_k=50,
    top_p=0.5,
    repetition_penalty=1.0
)

print("EXPERIMENT 4")
print("Parameter changed: Top-p")
print("Value: 0.5")
print()
print(output_4)
```

```python
set_seed(42)

output_5 = generate_text(
    prompt,
    temperature=1.0,
    top_k=50,
    top_p=1.0,
    repetition_penalty=1.5
)

print("EXPERIMENT 5")
print("Parameter changed: Repetition Penalty")
print("Value: 1.5")
print()
print(output_5)
```

## Comparison of Five Experiments

| Experiment | Parameter Changed | Value | Observation |
|---|---|---:|---|
| 1 | Temperature | 0.3 | The output was more consistent and focused. |
| 2 | Temperature | 1.2 | The output was more imaginative and random. |
| 3 | Top-k | 10 | Generation was limited to a smaller group of likely tokens. |
| 4 | Top-p | 0.5 | The output was more controlled by narrowing the probability choices. |
| 5 | Repetition Penalty | 1.5 | Repeated words and phrases were reduced. |

```python
print("=" * 70)
print("EXPERIMENT 1 - TEMPERATURE = 0.3")
print("=" * 70)
print(output_1)

print("\n")

print("=" * 70)
print("EXPERIMENT 2 - TEMPERATURE = 1.2")
print("=" * 70)
print(output_2)

print("\n")

print("=" * 70)
print("EXPERIMENT 3 - TOP-K = 10")
print("=" * 70)
print(output_3)

print("\n")

print("=" * 70)
print("EXPERIMENT 4 - TOP-P = 0.5")
print("=" * 70)
print(output_4)

print("\n")

print("=" * 70)
print("EXPERIMENT 5 - REPETITION PENALTY = 1.5")
print("=" * 70)
print(output_5)
```

## Analysis of the Five Experiments

A low temperature produced text that was generally more predictable and focused. With a higher temperature, the generated output became more varied and creative, but it could also lose some coherence.

Using a lower top-k value limited the model to fewer likely word choices, which made the output more controlled. In a similar way, lowering top-p restricted the cumulative probability range used for token selection.

The repetition penalty helped discourage the model from repeating the same words or phrases. Overall, temperature had a strong influence on randomness, while top-k and top-p affected the range of possible tokens considered during generation.
